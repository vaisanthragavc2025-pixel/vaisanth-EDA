```python
import os
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Step 1: Load data file
train_path = '/content/fraudTrain.csv'
df_train = pd.read_csv(train_path) if os.path.exists(train_path) else pd.DataFrame({'amt': [10.5, 50.0, 1200.0], 'age': [25, 45, 35]})

# Step 2: Binning (Fixed ranges and Quantiles)
bins = [-1, 20, 100, 1000, float('inf')]
labels = ['Micro', 'Standard', 'High', 'Extreme']

df_train['amt_labels'] = pd.cut(df_train['amt'], bins=bins, labels=labels)
df_train['amt_quantiles'] = pd.qcut(df_train['amt'], q=4, labels=['Q1', 'Q2', 'Q3', 'Q4'], duplicates='drop')

print("--- fraudTrain Binning Output ---")
print(df_train[['amt', 'amt_labels', 'amt_quantiles']])

# Step 3: Outlier Detection (Fixed limit and IQR)
limit_outliers_train = df_train[df_train['amt'] > 1000]
q1_tr, q3_tr = df_train['amt'].quantile(0.25), df_train['amt'].quantile(0.75)
iqr_tr = q3_tr - q1_tr
iqr_outliers_train = df_train[df_train['amt'] > (q3_tr + 3 * iqr_tr)]

print("\n--- fraudTrain Outlier Output ---")
print(f"Fixed Limit Outliers Count: {len(limit_outliers_train)}")
print(f"IQR Outliers Count: {len(iqr_outliers_train)}")

# Step 4: Data Transformation (Scaling)
features_train = df_train[['amt', 'age']]

mm_train = pd.DataFrame(MinMaxScaler().fit_transform(features_train), columns=['amt_minmax', 'age_minmax'])
std_train = pd.DataFrame(StandardScaler().fit_transform(features_train), columns=['amt_zscore', 'age_zscore'])
df_train_final = pd.concat([df_train, mm_train, std_train], axis=1)

print("\n--- fraudTrain Transformation Output ---")
print(df_train_final[['amt', 'amt_minmax', 'amt_zscore']])

```
