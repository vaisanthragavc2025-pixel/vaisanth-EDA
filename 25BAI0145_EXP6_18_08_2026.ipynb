## 1. Load Data and Initial Exploration

First, we'll load the `WineQT.csv` dataset into a pandas DataFrame and examine its basic properties, such as the first few rows, data types, and summary statistics. This will help us get an initial understanding of the data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load the dataset
df = pd.read_csv('/content/WineQT.csv')

print("First 5 rows of the dataset:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nDescriptive Statistics:")
display(df.describe())
```

## 2. Missing Values Check

It's crucial to check for any missing values before proceeding with the analysis. If missing values are present, appropriate handling (e.g., imputation or removal) will be necessary.

```python
print("Missing values per column:")
display(df.isnull().sum())
```

## 3. Multivariate Analysis: Correlation Matrix

To understand the relationships between different physicochemical properties and wine quality, we will compute and visualize the correlation matrix. This plot will highlight which features are strongly correlated with each other and with the target variable, 'quality'.

```python
plt.figure(figsize=(12, 10))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Wine Features and Quality')
plt.show()
```

## 4. Multivariate Analysis: Feature Distribution by Quality

We can use box plots to visualize the distribution of each physicochemical feature across different wine quality scores. This helps us see how the range and median of each feature vary with quality.

```python
features = df.columns.drop(['quality', 'Id'])
n_cols = 3
n_rows = int(np.ceil(len(features) / n_cols))

fig, axes = plt.subplots(n_rows, n_cols, figsize=(5 * n_cols, 4 * n_rows))
axes = axes.flatten()

for i, feature in enumerate(features):
    sns.boxplot(x='quality', y=feature, data=df, ax=axes[i], palette='viridis')
    axes[i].set_title(f'{feature} by Quality')
    axes[i].set_xlabel('Quality')
    axes[i].set_ylabel(feature)

# Remove any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

## 5. Multivariate Analysis: Pair Plot (Subset of Features)

To visualize the relationships between pairs of variables and their distribution, a pair plot is very useful. Since the dataset has many features, we'll select a subset of highly correlated features and 'quality' for a more readable plot. This can reveal trends, clusters, and the impact of 'quality' on these relationships.

```python
relevant_features = ['fixed acidity', 'volatile acidity', 'citric acid', 'alcohol', 'sulphates', 'quality']
sns.pairplot(df[relevant_features], hue='quality', palette='viridis', diag_kind='kde')
plt.suptitle('Pair Plot of Selected Features by Quality', y=1.02) # Adjust suptitle position
plt.show()
```

## 6. Multivariate Analysis: Principal Component Analysis (PCA)

PCA is a dimensionality reduction technique that can help visualize high-dimensional data in 2D or 3D. We will apply PCA to the standardized features and visualize the first two principal components, colored by wine quality, to see if quality groups are separable in a lower-dimensional space.

```python
# Prepare the data for PCA (exclude 'quality' and 'Id')
X = df.drop(['quality', 'Id'], axis=1)
y = df['quality']

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply PCA
pca = PCA(n_components=2) # Reduce to 2 principal components
X_pca = pca.fit_transform(X_scaled)

# Create a DataFrame for the PCA results
pca_df = pd.DataFrame(data=X_pca, columns=['Principal Component 1', 'Principal Component 2'])
pca_df['Quality'] = y

# Plot the PCA results
plt.figure(figsize=(10, 8))
sns.scatterplot(x='Principal Component 1', y='Principal Component 2', hue='Quality', palette='viridis', data=pca_df, s=100, alpha=0.8)
plt.title('PCA of Wine Features by Quality')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True)
plt.show()

print("Explained variance ratio by each principal component:")
print(pca.explained_variance_ratio_)
```

## 7. Quality Distribution

Let's visualize the distribution of wine quality scores using a count plot to see how many wines fall into each quality category.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/WineQT.csv')

plt.figure(figsize=(10, 6))
df["quality"].value_counts().nlargest(30).plot(kind='bar', color="teal")
plt.title("Distribution of Wine Quality Scores")
plt.ylabel('Number of Wines')
plt.xlabel('Quality Score')
plt.tight_layout()
plt.show()
```
