```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Iris dataset
iris_df = pd.read_csv('/content/Iris.csv')

# Display the first 5 rows of the DataFrame
display(iris_df.head())
```

```python
# Display basic information about the dataset
print(iris_df.info())
```

### Data Distribution Analysis

Let's visualize the distribution of each numerical feature using histograms and check for species counts.

```python
# Histograms for each feature
iris_df.hist(edgecolor='black', figsize=(10, 8))
plt.suptitle('Distribution of Iris Features', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Count of each Iris species
plt.figure(figsize=(7, 5))
sns.countplot(x='Species', data=iris_df, palette='viridis', hue='Species', legend=False)
plt.title('Count of Iris Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.show()
```

### Relationships Between Variables

Now, let's explore the relationships between different features and how they vary across species using a pair plot.

```python
# Pair plot to show relationships between all features, colored by species
sns.pairplot(iris_df, hue='Species', markers=['o', 's', 'D'])
plt.suptitle('Pair Plot of Iris Features by Species', y=1.02)
plt.show()
```

### Box Plots for Feature Comparison

Box plots provide another view of feature distributions and comparisons across species, highlighting median, quartiles, and outliers.

```python
# Box plots for each feature compared across species
plt.figure(figsize=(15, 10))

# Select only the relevant measurement features, excluding 'Id'
measurement_features = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']

for i, feature in enumerate(measurement_features):
    plt.subplot(2, 2, i + 1) # Arrange in 2x2 grid
    sns.boxplot(x='Species', y=feature, data=iris_df, palette='muted')
    plt.title(f'Distribution of {feature} by Species')
plt.tight_layout()
plt.show()
```
