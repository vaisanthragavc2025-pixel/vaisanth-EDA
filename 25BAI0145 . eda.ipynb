```python
import pandas as pd

# =========================================================================
# PREPARATION: RECONSTRUCTING SAMPLE DATASETS FROM THE UPLOADED DOCUMENT
# =========================================================================

# Creating DataFrame 1 from Page 1 (Top records)
data_p1 = {
    'Region': ['Middle East and North Africa', 'North America', 'Middle East and North Africa', 'Asia'],
    'Country': ['Libya', 'Canada', 'Libya', 'Japan'],
    'Item Type': ['Cosmetics', 'Vegetables', 'Baby Food', 'Cereal'],
    'Sales Channel': ['Offline', 'Online', 'Offline', 'Offline'],
    'Units Sold': [8446, 3018, 1517, 3322],
    'UnitPrice': [437.20, 154.06, 255.28, 205.70]
}
df_sales1 = pd.DataFrame(data_p1)

# Creating DataFrame 2 from Page 2 (Continuing records)
data_p2 = {
    'Region': ['Europe', 'Europe', 'Sub-Saharan Africa', 'Asia'],
    'Country': ['Ukraine', 'Croatia', 'Madagascar', 'Malaysia'],
    'Item Type': ['Cosmetics', 'Beverages', 'Fruits', 'Snacks'],
    'Sales Channel': ['Online', 'Online', 'Online', 'Offline'],
    'Units Sold': [2484, 1629, 213, 897],
    'UnitPrice': [81.73, 152.58, 437.20, 47.45]
}
df_sales2 = pd.DataFrame(data_p2)


# =========================================================================
# 1. CONCATENATION OF SALES DATAFRAMES
# =========================================================================

# Vertical Concatenation: Stacking data from different pages over each other
vertical_concat = pd.concat([df_sales1, df_sales2], ignore_index=True)
print("--- 1. Vertical Concatenation (Combined Pages Data) ---")
print(vertical_concat[['Country', 'Item Type', 'Units Sold']])

# Horizontal Concatenation: Side-by-side comparison binding
horizontal_concat = pd.concat([df_sales1, df_sales2], axis=1)
print("\n--- Horizontal Concatenation (Side-by-Side Binding) ---")
print(horizontal_concat)


# =========================================================================
# 2. MERGING DATAFRAMES (JOINS ON COUNTRY)
# =========================================================================

# Subset A: Pricing information catalog
df_prices = pd.DataFrame({
    'Country': ['Libya', 'Canada', 'Japan', 'Ukraine', 'Croatia', 'Brazil'],
    'LocalTaxRate': [0.05, 0.12, 0.08, 0.20, 0.25, 0.15]
})

# Subset B: Shipping channels status
df_shipping = pd.DataFrame({
    'Country': ['Libya', 'Canada', 'Japan', 'Ukraine', 'Madagascar', 'India'],
    'ShippingStatus': ['Delivered', 'In-Transit', 'Delivered', 'Pending', 'Delivered', 'Cancelled']
})

# Inner Join: Keeps countries that exist in BOTH subsets (Libya, Canada, Japan, Ukraine)
inner_join = pd.merge(df_prices, df_shipping, on='Country', how='inner')
print("\n--- 2. Inner Join Results ---")
print(inner_join)

# Left Join: Keeps all records from df_prices, fills missing shipping statuses with NaN
left_join = pd.merge(df_prices, df_shipping, on='Country', how='left')
print("\n--- Left Join Results ---")
print(left_join)

# Outer Join: Keeps every unique country record across both sets
outer_join = pd.merge(df_prices, df_shipping, on='Country', how='outer')
print("\n--- Outer Join Results ---")
print(outer_join)


# =========================================================================
# 3. ADDING A NEW COLUMN (FINANCIAL FORMULAS FROM PAGE 20)
# =========================================================================

# Calculating Total Revenue based on your document logic (Units Sold * UnitPrice)
vertical_concat['TotalRevenue'] = vertical_concat['Units Sold'] * vertical_concat['UnitPrice']
print("\n--- 3. Sales Data with Calculated Total Revenue ---")
print(vertical_concat[['Country', 'Item Type', 'Units Sold', 'UnitPrice', 'TotalRevenue']])


# =========================================================================
# 4. DESCRIPTIVE STATISTICS
# =========================================================================

print("\n--- 4. Descriptive Statistics of Key Metrics ---")
print(vertical_concat[['Units Sold', 'UnitPrice', 'TotalRevenue']].describe())


# =========================================================================
# 5. VALUE COUNTS (CATEGORICAL SUMMARIES)
# =========================================================================

print("\n--- 5. Order Distributions by Region ---")
print(vertical_concat['Region'].value_counts())

print("\n--- Channels Distribution ---")
print(vertical_concat['Sales Channel'].value_counts())


# =========================================================================
# 6. HIERARCHICAL INDEXING (STACK/UNSTACK ON REGIONAL METRICS)
# =========================================================================

# Building a regional performance matrix frame
summary_matrix = vertical_concat.groupby(['Region', 'Item Type'])['Units Sold'].sum().unstack(fill_value=0)
print("\n--- 6. Original Reshaped Matrix ---")
print(summary_matrix)

# Stacking back to multi-index series
stacked_sales = summary_matrix.stack()
print("\n--- Stacked Regional Multi-Index Data ---")
print(stacked_sales.head(10))
```
