```python
import pandas as pd
import numpy as np # Import numpy for np.arange

# Dataframe 1: Online Store Customers
df1_online = pd.DataFrame({
    'CustomerID': np.arange(1, 4), # Added values for CustomerID
    'OnlineSpend': [150, 230, 45]
})

# Dataframe 2: Physical Store Customers
df2_physical = pd.DataFrame({
    'CustomerID': np.arange(3, 6), # Added values for CustomerID
    'InStoreSpend': [310, 75, 410]
})

# Complete master datasets via concatenation
df_online_all = pd.concat([df1_online], ignore_index=True)
df_physical_all = pd.concat([df2_physical], ignore_index=True)

# Option 1: Concatenation by Axis (Side-by-side)
df_axis = pd.concat([df_online_all, df_physical_all], axis=1)
display(df_axis)

# Option 2: Inner Join (Omnichannel Customers)
df_inner = df_online_all.merge(df_physical_all, how='inner')
display(df_inner)

# Option 3: Left Join (All Online Customers)
df_left = df_online_all.merge(df_physical_all, how='left')
display(df_left)

# Option 4: Right Join (All Physical Store Customers)
df_right = df_online_all.merge(df_physical_all, how='right')
display(df_right)

# Sales Registry Dataset Replacement
sales_data = {
    'TransactionID': np.arange(1001, 1006), # Added values for TransactionID
    'Company': ['Logitech', 'Apple', 'Logitech', 'Samsung', 'Apple'],
    'UnitPrice': [50, 1200, 60, 300, 1500], # Added values for UnitPrice
    'Quantity': [3, 1, 2, 4, 2]
}
df_sales = pd.DataFrame(sales_data)
display(df_sales.head(10))

# Feature Engineering: Total Price Calculation
df_sales['TotalPrice'] = df_sales['UnitPrice'] * df_sales['Quantity']
display(df_sales.head(10))

# Value Counts and Summary Statistics
display(df_sales['Company'].value_counts())
display(df_sales.describe())
```
